```python
import math

def minimax(depth, index, is_max, values, max_depth):
    # Leaf node
    if depth == max_depth:
        return values[index]

    if is_max:
        return max(
            minimax(depth + 1, index * 2, False, values, max_depth),
            minimax(depth + 1, index * 2 + 1, False, values, max_depth)
        )
    else:
        return min(
            minimax(depth + 1, index * 2, True, values, max_depth),
            minimax(depth + 1, index * 2 + 1, True, values, max_depth)
        )


# Example usage (edit values based on your tree)
values = [-4, -5, 2, 3, 4, 5, 6, 7, 8, 0, 1, 8]

max_depth = math.floor(math.log(len(values), 2))

result = minimax(0, 0, True, values, max_depth)

print("Minimax Result:", result)
```

```python
import math

def alphabeta(depth, index, is_max, values, max_depth, alpha, beta):
    # Leaf node
    if depth == max_depth:
        return values[index]

    if is_max:
        best = -math.inf

        left = alphabeta(depth + 1, index * 2, False, values, max_depth, alpha, beta)
        best = max(best, left)
        alpha = max(alpha, best)

        if beta <= alpha:
            return best  # prune

        right = alphabeta(depth + 1, index * 2 + 1, False, values, max_depth, alpha, beta)
        best = max(best, right)
        alpha = max(alpha, best)

        return best

    else:
        best = math.inf

        left = alphabeta(depth + 1, index * 2, True, values, max_depth, alpha, beta)
        best = min(best, left)
        beta = min(beta, best)

        if beta <= alpha:
            return best  # prune

        right = alphabeta(depth + 1, index * 2 + 1, True, values, max_depth, alpha, beta)
        best = min(best, right)
        beta = min(beta, best)

        return best


# Example usage
values = [-4, -5, 2, 3, 4, 5, 6, 7, 8, 0, 1, 8]

max_depth = math.floor(math.log(len(values), 2))

result = alphabeta(0, 0, True, values, max_depth, -math.inf, math.inf)

print("Alpha-Beta Result:", result)
```
